## 2. Signal Configuration

### Numeric controls

| Control Name | Purpose |
|---|---|
| `Basic freq (Hz)` | Basic pulse frequency |
| `Basic amp` | Basic pulse amplitude |
| `Duty cycle %` | Basic pulse duty cycle |
| `Gauss amp` | Gaussian amplitude |
| `Gauss freq (Hz)` | Gaussian frequency |
| `Gauss atten (dB)` | Gaussian attenuation |
| `Gauss Nor. Band` | Gaussian normalized bandwidth |
| `Chirp amp` | Chirp amplitude |
| `Chirp f1` | Chirp start frequency |
| `Chirp f2` | Chirp end frequency |
| `X Auto Distance` | Grid scan X distance |
| `Y Auto Distance` | Grid scan Y distance |
| `X Distance` | Manual move X |
| `Y Distance` | Manual move Y |
| `Z Distance` | Manual move Z |
| `Speed (mm/s)` | Movement speed |

```python
# set_numeric_control: Basic pulse parameters
tool_set_numeric_control(control_name="Basic freq (Hz)", value=1000.0)
tool_set_numeric_control(control_name="Basic amp", value=6.0)
tool_set_numeric_control(control_name="Duty cycle %", value=50.0)
```

```python
# set_numeric_control: Gaussian pulse parameters
tool_set_numeric_control(control_name="Gauss freq (Hz)", value=1500.0)
tool_set_numeric_control(control_name="Gauss amp", value=3.0)
tool_set_numeric_control(control_name="Gauss atten (dB)", value=10.0)
tool_set_numeric_control(control_name="Gauss Nor. Band", value=0.5)
```

```python
# set_numeric_control: Chirp pulse parameters
tool_set_numeric_control(control_name="Chirp amp", value=5.5)
tool_set_numeric_control(control_name="Chirp f1", value=500.0)
tool_set_numeric_control(control_name="Chirp f2", value=2500.0)
```

```python
# get_numeric_indicator: Read a value from the front panel
tool_get_numeric_indicator(indicator_name="Basic freq (Hz)")
```
