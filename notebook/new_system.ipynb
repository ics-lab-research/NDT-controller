```python
from src.tools.labview_controller import *
```

```python
# use timestamp to avoid overwriting existing files
from datetime import datetime
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
filename = f"scan_{timestamp}.tdms"

#=================================================================================


tool_connect_to_vi(vi_path=r"D:\Dat\Pulsed Eddy Current - LLM\MAIN - PECT - Autonics - Pulses - encoder  - LLM.vi")
tool_select_tab(tab_control_name="Measurement Modes", tab=1) # chang to scan mode

# set parameters
tool_set_numeric_control(control_name="X Auto Distance", value=5)
tool_set_numeric_control(control_name="Y Auto Distance", value=6)

tool_set_numeric_control(control_name="X Step", value=1)
tool_set_numeric_control(control_name="Y Step", value=1)


tool_set_numeric_control(control_name="Speed (mm/s)", value=5)

# BUG: set parameters in Pulse Types tab not working


tool_run_open_vi(run_vi=True)
tool_handle_save_dialog(save_directory=r"D:\Dat\NDT-controller\EXPERIMENTS\data", filename=filename)

tool_connect_to_vi(vi_path=r"D:\Dat\Pulsed Eddy Current - LLM\MAIN - PECT - Autonics - Pulses - encoder  - LLM.vi")
tool_set_button_state("Scan", state=True)  # Start scan
```

```python
# check is done
tool_connect_to_vi(vi_path=r"D:\Dat\Pulsed Eddy Current - LLM\MAIN - PECT - Autonics - Pulses - encoder  - LLM.vi")
scan_status = tool_get_numeric_indicator("Completed")
scan_status
```

```python
is_done = bool(float(scan_status.output['value']))
```

```python
if is_done:
    print("Scan completed.")
else:
    print("Scan still in progress.")
```

```python
# when scan completed, go home position
if is_done: 
    pass
```

```python
x_distance_indicator = tool_get_numeric_indicator(indicator_name="X Auto Distance")
x_distance = float(x_distance_indicator.output["value"])

y_distance_indicator = tool_get_numeric_indicator(indicator_name="Y Auto Distance")
y_distance = float(y_distance_indicator.output["value"])
```

```python
tool_set_numeric_control(control_name="X Distance", value=x_distance)
tool_set_numeric_control(control_name="Y Distance", value=y_distance)
```
