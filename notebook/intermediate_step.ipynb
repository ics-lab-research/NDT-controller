```python
import time

VI_PATH = r"D:\PECT LLM\MAIN - PECT - Autonics - Pulses - encoder_TCP1.vi"
SAVE_DIR = r"D:\GitHub\river\data"
SAVE_NAME = "scan_test5"

# --- Connect and configure scan ---
tool_connect_to_vi(vi_path=VI_PATH)
tool_select_tab("Tab Control 2", 1)  # Switch to scan mode

# Scan ranges
tool_set_numeric_control("xl", 5.0)
tool_set_numeric_control("yl", 5.0)

# Step sizes
tool_set_numeric_control("Step X", 1.0)
tool_set_numeric_control("Step Y", 1.0)

# Start VI in open mode
tool_run_open_vi(True)

# Save dialog handling
tool_handle_save_dialog(save_directory=SAVE_DIR, filename=SAVE_NAME)

# Set home before scan
tool_connect_to_vi(vi_path=VI_PATH)
tool_set_button_state("sethome", True)
time.sleep(1)

# Start Scan
tool_set_button_state("Scan", True)

print("Scan started... waiting for completion")

# --- Wait for scan to complete ---
while True:
    tool_connect_to_vi(vi_path=VI_PATH)
    status = tool_get_numeric_indicator("Completed")

    if status["value"]:
        print("Scan completed!")
        break

    time.sleep(1)  # poll every second

# --- Return home and quit ---
tool_set_button_state("gohome", True)
time.sleep(5)

tool_stop_vi(close_labview=True)
print("Returned home and closed LabVIEW.")
```
